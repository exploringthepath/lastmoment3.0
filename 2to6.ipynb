```python
# 2
import tensorflow as tf                  # Import TensorFlow for deep learning
import matplotlib.pyplot as plt          # Import Matplotlib to plot graphs
from tensorflow.keras import Sequential   # Create a model with layers in order
from tensorflow.keras.layers import Flatten, Dense  # Import the required layers
from tensorflow.keras.datasets import mnist          # Import the MNIST dataset

(x_train, y_train), (x_test, y_test) = mnist.load_data()  # Load training and testing data

x_train = x_train / 255.0                # Normalize training pixels between 0 and 1
x_test = x_test / 255.0                  # Normalize testing pixels between 0 and 1

model = Sequential([                     # Create the neural network
    Flatten(input_shape=(28, 28)),        # Convert each 28x28 image into 784 values
    Dense(128, activation="relu"),        # Hidden layer with 128 neurons and ReLU
    Dense(10, activation="softmax")       # Output layer for the 10 digit classes
])

model.compile(                             # Configure the model before training
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.01),  # Use SGD to update weights
    loss="sparse_categorical_crossentropy",                 # Measure prediction errors
    metrics=["accuracy"]                                    # Track prediction accuracy
)

history = model.fit(                       # Train the neural network
    x_train, y_train,                       # Provide images and their correct labels
    epochs=10,                               # Train for 5 complete passes
    batch_size=128,                         # Process 128 images in each batch
    validation_split=0.1                     # Reserve 10% of training data for validation
)

loss, accuracy = model.evaluate(x_test, y_test)  # Evaluate using unseen test images
print("Test accuracy:", accuracy * 100, "%")     # Display test accuracy as a percentage

plt.plot(history.history["loss"], label="Training Loss")  # Plot training loss
plt.plot(history.history["val_loss"], label="Validation Loss")  # Plot validation loss
plt.legend()                              # Display the graph labels
plt.show()                                # Display the loss graph

plt.plot(history.history["accuracy"], label="Training Accuracy")  # Plot training accuracy
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")  # Plot validation accuracy
plt.legend()                              # Display the graph labels
plt.show()                                # Display the accuracy graph
```

```python
# 4
# 1. Import required libraries
import numpy as np                         # For mathematical calculations
import matplotlib.pyplot as plt            # For displaying graphs and images
import tensorflow as tf                    # For building neural networks
from tensorflow.keras import layers, Model # For creating the encoder and decoder
from tensorflow.keras.datasets import mnist # For loading the MNIST dataset

# 2. Load and preprocess the dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data() # Load handwritten digit images

x_train = x_train.astype("float32") / 255.0 # Normalize training pixels between 0 and 1
x_test = x_test.astype("float32") / 255.0   # Normalize testing pixels between 0 and 1

x_train = x_train.reshape(-1, 784)          # Flatten 28x28 images into 784 features
x_test = x_test.reshape(-1, 784)            # Flatten test images into 784 features

x_train_normal = x_train[y_train == 0]      # Select only digit 0 as normal training data
y_test_anomaly = (y_test != 0).astype(int)  # Mark digit 0 as normal and other digits as anomalies

# 3. Build the Encoder
input_dim = 784                             # Number of input features
latent_dim = 32                             # Compressed representation size

encoder_input = layers.Input(shape=(784,))  # Define encoder input
x = layers.Dense(128, activation="relu")(encoder_input) # First hidden layer
x = layers.Dense(64, activation="relu")(x)  # Second hidden layer
encoded = layers.Dense(32, activation="relu")(x) # Create compressed representation

encoder = Model(encoder_input, encoded)     # Create encoder model

# 4. Build the Decoder
decoder_input = layers.Input(shape=(32,))   # Accept compressed input
x = layers.Dense(64, activation="relu")(decoder_input) # First decoder layer
x = layers.Dense(128, activation="relu")(x) # Second decoder layer
decoded = layers.Dense(784, activation="sigmoid")(x) # Reconstruct original 784 features

decoder = Model(decoder_input, decoded)     # Create decoder model

# Combine encoder and decoder
auto_input = layers.Input(shape=(784,))     # Define complete autoencoder input
auto_output = decoder(encoder(auto_input))  # Connect encoder output to decoder
autoencoder = Model(auto_input, auto_output) # Create complete autoencoder

# 5. Compile and train the autoencoder
autoencoder.compile(optimizer="adam", loss="mse", metrics=["mae"]) # Configure model

history = autoencoder.fit(
    x_train_normal, x_train_normal,          # Input and expected output are the same
    epochs=10,                               # Train for 10 complete passes
    batch_size=128,                          # Process 128 images at a time
    validation_split=0.1,                    # Use 10% data for validation
    verbose=1                                # Display training progress
)

# 6. Calculate reconstruction error
reconstructed = autoencoder.predict(x_test, verbose=0) # Reconstruct test images

errors = np.mean((x_test - reconstructed) ** 2, axis=1) # Calculate MSE for each image

normal_reconstructed = autoencoder.predict(x_train_normal, verbose=0) # Reconstruct normal training images
normal_errors = np.mean((x_train_normal - normal_reconstructed) ** 2, axis=1) # Calculate normal errors

threshold = np.percentile(normal_errors, 95) # Set threshold at 95th percentile

y_pred = (errors > threshold).astype(int)    # Predict anomalies using threshold

# 7. Evaluate the model
accuracy = np.mean(y_pred == y_test_anomaly) # Calculate overall accuracy

tp = np.sum((y_pred == 1) & (y_test_anomaly == 1)) # Count true positives
fp = np.sum((y_pred == 1) & (y_test_anomaly == 0)) # Count false positives
fn = np.sum((y_pred == 0) & (y_test_anomaly == 1)) # Count false negatives

precision = tp / (tp + fp + 1e-8)           # Calculate precision
recall = tp / (tp + fn + 1e-8)              # Calculate recall

print("Threshold:", threshold)               # Display anomaly threshold
print("Accuracy:", accuracy * 100)           # Display accuracy percentage
print("Precision:", precision)               # Display precision
print("Recall:", recall)                     # Display recall

# 8. Plot training and validation loss
plt.plot(history.history["loss"], label="Training Loss") # Plot training loss
plt.plot(history.history["val_loss"], label="Validation Loss") # Plot validation loss
plt.xlabel("Epochs")                          # Label X-axis
plt.ylabel("MSE Loss")                        # Label Y-axis
plt.legend()                                  # Display graph labels
plt.title("Autoencoder Training Loss")        # Add graph title
plt.show()                                    # Display graph

# 9. Display 2 normal and 2 anomalous examples
for label, name in [(0, "Normal"), (1, "Anomalous")]: # Process both categories
    indices = np.where(y_test_anomaly == label)[0][:2] # Select 2 images

    plt.figure(figsize=(6, 3))                # Create figure

    for i, idx in enumerate(indices):         # Process selected images
        plt.subplot(2, 2, i + 1)              # Position original image
        plt.imshow(x_test[idx].reshape(28, 28), cmap="gray") # Display original image
        plt.title("Original")                  # Add title
        plt.axis("off")                        # Hide axes

        plt.subplot(2, 2, i + 3)              # Position reconstructed image
        plt.imshow(reconstructed[idx].reshape(28, 28), cmap="gray") # Display reconstruction
        plt.title("Reconstructed")             # Add title
        plt.axis("off")                        # Hide axes

    plt.suptitle(name + " Images")             # Add category title
    plt.tight_layout()                         # Adjust spacing
    plt.show()                                 # Display images
```

```python
# 5
# 1. Import libraries
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Sequential

# 2. Prepare sample sentences
sentences = [
    "the cat sits on the mat",
    "the dog sits on the rug",
    "the cat eats fish",
    "the dog eats meat",
    "a cat likes fish",
    "a dog likes meat",
    "the cat likes the mat",
    "the dog likes the rug"
]

# 3. Tokenization and vocabulary creation
words = [s.lower().split() for s in sentences]  # Convert sentences into lowercase words
vocab = sorted(set(w for s in words for w in s))  # Get unique words

word_to_id = {w:i+1 for i,w in enumerate(vocab)}  # Assign IDs starting from 1
id_to_word = {i:w for w,i in word_to_id.items()}  # Convert IDs back into words
vocab_size = len(vocab) + 1  # Reserve 0 for padding

print("Vocabulary:", vocab)
print("Vocabulary size:", len(vocab))
print("Word to ID:", word_to_id)

# 4. Generate context and target pairs
window = 2  # Consider two words on each side
contexts, targets = [], []  # Store context and target words

for sentence in words:  # Process each sentence
    ids = [word_to_id[w] for w in sentence]  # Convert words into IDs

    for i in range(len(ids)):  # Visit every word
        context = ids[max(0,i-window):i] + ids[i+1:i+window+1]  # Get surrounding words

        if context:  # Ignore empty contexts
            contexts.append(context)  # Store context words
            targets.append(ids[i])  # Store target word

# 5. Pad context sequences
X = tf.keras.utils.pad_sequences(contexts, maxlen=4, padding='post', value=0)  # Make inputs equal length
y = np.array(targets)  # Convert targets into an array

print("Number of training examples:", len(y))

for i in range(5):  # Display five training examples
    print("Context:", [id_to_word[w] for w in contexts[i]],
          "-> Target:", id_to_word[targets[i]])

# 6. Build CBOW model
model = Sequential([
    layers.Embedding(vocab_size, 16, mask_zero=True),  # Convert word IDs into vectors
    layers.GlobalAveragePooling1D(),  # Average context word vectors
    layers.Dense(vocab_size, activation='softmax')  # Predict target word probabilities
])

# 7. Compile model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])  # Configure model learning

# 8. Train model
history = model.fit(X, y, epochs=150, verbose=0)  # Train using context-target pairs

print("Training completed!")
print("Final training accuracy:",
      round(history.history['accuracy'][-1] * 100, 2), "%")

# 9. Predict target word
test_words = ["cat", "on", "mat"]  # Give surrounding context words
test_ids = [word_to_id[w] for w in test_words]  # Convert words into IDs

test_input = tf.keras.utils.pad_sequences(
    [test_ids], maxlen=4, padding='post', value=0)  # Prepare model input

probabilities = model.predict(test_input, verbose=0)[0]  # Get prediction probabilities
top_ids = np.argsort(probabilities)[::-1][:5]  # Select top five predictions

print("\nContext words:", test_words)
print("Predicted target words:")

for i in top_ids:  # Display top five predicted words
    print(id_to_word[i], "probability =", round(probabilities[i], 4))

# 10. Display actual vs predicted words
print("\nSample training predictions:")

for i in range(5):  # Check five training examples
    prediction = model.predict(X[i:i+1], verbose=0)[0]  # Predict target word
    predicted_id = np.argmax(prediction)  # Select highest probability word

    print("Context:", [id_to_word[w] for w in contexts[i]],
          "| Actual:", id_to_word[targets[i]],
          "| Predicted:", id_to_word[predicted_id])
```

```python

# 1. Import libraries
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import cifar10

# 2. Load CIFAR-10 dataset
(x_train, y_train), (x_test, y_test) = cifar10.load_data()
y_train = y_train.flatten()
y_test = y_test.flatten()

# Use a smaller dataset to save time
x_train, y_train = x_train[:10000], y_train[:10000]
x_test, y_test = x_test[:2000], y_test[:2000]

# 3. Load pre-trained MobileNetV2
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(96, 96, 3),
    include_top=False,
    weights="imagenet"
)

# Freeze pre-trained layers
base_model.trainable = False

# 4. Add custom classifier
inputs = layers.Input(shape=(32, 32, 3))
x = layers.Resizing(96, 96)(inputs)
x = tf.keras.applications.mobilenet_v2.preprocess_input(x)

x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dense(128, activation="relu")(x)
outputs = layers.Dense(10, activation="softmax")(x)

model = models.Model(inputs, outputs)

# 5. Compile the model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# 6. Train the custom classifier
# model.fit(x_train, y_train, epochs=5, batch_size=32)

# Train the custom classifier and save its history
history1 = model.fit(x_train, y_train, epochs=5, batch_size=32, validation_split=0.2)

# Fine-tune the model and save its history
history2 = model.fit(x_train, y_train, epochs=3, batch_size=32, validation_split=0.2)


# 7. Fine-tune the model
base_model.trainable = True

# Freeze earlier layers; train only upper layers
for layer in base_model.layers[:-30]:
    layer.trainable = False

# Recompile with a low learning rate
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.00001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Train again

# Train the custom classifier and save its history
history1 = model.fit(x_train, y_train, epochs=5, batch_size=32, validation_split=0.2)

# Fine-tune the model and save its history
history2 = model.fit(x_train, y_train, epochs=3, batch_size=32, validation_split=0.2)


# 8. Evaluate the model
loss, accuracy = model.evaluate(x_test, y_test)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy * 100, "%")

import matplotlib.pyplot as plt                 # Import plotting library

# Combine accuracy values from both training stages
train_acc = history1.history["accuracy"] + history2.history["accuracy"]
val_acc = history1.history["val_accuracy"] + history2.history["val_accuracy"]

# Plot training and validation accuracy
plt.plot(train_acc, label="Training Accuracy")
plt.plot(val_acc, label="Validation Accuracy")
plt.axvline(4.5, linestyle="--", label="Fine-tuning starts")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Transfer Learning Accuracy")
plt.legend()
plt.show()

```
